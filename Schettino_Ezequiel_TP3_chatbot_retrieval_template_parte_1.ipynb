# ChatBot con TF-IDF y similitud del coseno.

#**Prompt** Explicame cada parte de este código.

```python
# Realizamos el código para la similitud del coseno entre la consulta y la lista de preguntas como función para que sea reutilizable.
def responder_pregunta(consulta_usuario):
    # Transforma la consulta del usuario en un vector numérico usando el modelo TF-IDF.
    consulta_vec = vectorizer.transform([consulta_usuario])

    # Calcula la similitud del coseno entre la consulta del usario y la matrix de la lista de preguntas.
    # Lo convierte en una lista plana, para que pueda encontrarse más fácilmente el valor que más se aproxima (argmax., por ejemplo)
    similitudes = cosine_similarity(consulta_vec, tfidf_matrix_preg).flatten()
    print(similitudes)

    # Acá es útil, entonces, la lista plana.
    indice_mejor_coincidencia = similitudes.argmax()

    # Umbral mínimo de confianza. Es un umbral bajo, lo hicimos exigente.
    if similitudes[indice_mejor_coincidencia] < 0.2:
        return "Lo siento, no comprendí tu consulta. ¿Podrías reformularla?"
    print(indice_mejor_coincidencia)

    # Retornamos la mejor coincidencia.
    return respuestas[indice_mejor_coincidencia]
```

```python
# Función para el inicio de la primera versión de EsTeFI.
def iniciar_chatbot():
    print("¡EsTeFI lista! Escribe 'chau, gracias' para terminar.\n")

    # Ciclo repetitivo para las entradas con una frase de corte.
    while True:
        entrada = input("Tú: ")
        if entrada.lower() == 'chau, gracias':
            print("¡Hasta luego!")
            break

        # Llama a tu función ya creada y le asignamos como parámetro la entrada o consulta del usuario. La variable es 'resultado'.
        resultado = responder_pregunta(entrada)
        print(f"EsTeFI: {resultado}\n" + "-"*100)
```

# Chatbot 2: Utilizando word embeddings.

# Usando el código de la clase, con la ayuda de Gemini, lo hice una función.

```python
# Cargar el modelo de spaCy
nlp = spacy.load("es_core_news_md")

# Vectorizar las preguntas previamente (optimización)
X_embeddings = np.array([nlp(p).vector for p in preguntas])

# Función para responder preguntas con embeddings y umbral mínimo
def responder_pregunta_embeddings(consulta_usuario):
    vector_consulta = nlp(consulta_usuario).vector.reshape(1, -1)
    similitudes = cosine_similarity(vector_consulta, X_embeddings).flatten()

    indice_mejor_coincidencia = similitudes.argmax()

    # Umbral mínimo de confianza
    if similitudes[indice_mejor_coincidencia] < 0.2:
        return "Lo siento, no comprendí tu consulta. ¿Podrías reformularla?"

    return respuestas[indice_mejor_coincidencia]

# Función para iniciar el chat interactivo
def iniciar_chatbot_2():
    print("¡Chatbot con Embeddings listo! Escribe 'chau, gracias' para terminar.\n")

    while True:
        entrada = input("Tú: ")
        if entrada.lower() == 'chau, gracias':
            print("Chatbot: ¡Hasta luego!")
            break

        resultado = responder_pregunta_embeddings(entrada)
        print(f"Chatbot: {resultado}\n" + "-"*100)
```

Si tomamos el la estadística de nuestras consultas hechas en los chatbots, las cuales fueron similares pero no iguales a la lista de consultas asignadas en el bag of words (la lista 'preguntas' que le dimos), el resultado es el siguiente:

Entendemos que esta mayor tasa de aciertos a favor de EsTeFI está en el orden que el TF-IDF realiza la cuenta correspondiente sólo entre la entrada del usuario y la lista de preguntas y de acuerdo a la cercanía de estos cosenos, me lo deriva a la pregunta que más se le acerca (por más que esta cercanía sea 0.01 pero si es la mayor enseguida me saca el output correspondiente a ese número de respuesta). Acá compara palabra por palabra dentro del entorno que le doy yo, sea entrada del usuario con la lista de preguntas y me lo deriva a la lista de respuestas o a la condición de corte que corresponda para sacar el mensaje estipulado, en este caso "Lo siento, no entiendo su pregunta...".

Por el lado del Word Embedding, en este caso usamos el 'es_core_news_md', este ya viene con una bag of words de millones de palabras, este específicamente tiene alrededor de 15 millones de términos en español y una cantidad de 300 dimensiones (a diferencia de EsTeFI que maneja 99 dimensiones que son las palabras distintas con las que yo la embebí).
Entonces, las word embedding trabajan de una manera diferente, ya que compara el insight del usuario con sus 15 o 16 millones de términos y elabora un vector que sea similar (el cual será, conceptual y semanticamente, similar al entendimiento nuestro) y en base a este vector o esta frase que se creó, la compara con nuestra lista de entrada (preguntas). Esta comparación es la similitud del coseno correspondiente. Si este está por arriba del umbral de similitud que le puse, en este caso 0.2 (bajo, que sea muy similar digamos), ahí sí me la corresponde con la salida (la respuesta de la lista correspondiente), pero si está por debajo de ese umbral de cercanía, me va a la condición de salida "Los siento, no entiendo su pregunta...".
En este caso, usando 'es_core_news_md' (de tamaño medio) o este forma de construcción del chatbot, las palabras comunes, pueden hacer que las distancias se hagan más grandes, pero a palabras muy específicas (dialectos por ejemplo) puede ser que acerquen los vectores o sus cosenos correspondientes, ya que word embeddings con su gran universo de vocabulario, termina realizando una conceptualización semántica superior a mi TF IDF.
En nuestro caso, TF-IDF funcionó aproximadamente un 30% mejor ya que tenía mayor cantidad de palabras que estaban en mi bag of word de 99 palabras y tuvieron más peso que con el 'es_core_news_md' que tenía un universo de 16 millones de palabras y y 300 dimensiones, si no había palabras muy específicas, las distancias se alejaban (y sobre todo teniendo en cuenta el umbral que le puse, 0.2).

Una prueba visual de esta diferencia entre magnitudes, es el gráfico que realizamos de ambos sistemas o herramientes: Mientras que las frases de TF-IDF están más dispersas (universo mucho más pequeño), las mismas frases en el word embeddings está mucho más cerca ya que realizan en su enorme universo una conceptalización semántica, es decir, entiendo que saben que hablan, en general, de mismo tema, una atención universitaria.

Para entornos reducidos y específicos, TF-IDF es superior, pero para un análisis conceptual y semántico, word embeddings siempre será mucho más eficaz.
